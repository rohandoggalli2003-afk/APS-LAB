```python
#problem 1
import random
n=100
tosses=[random.choice(["Heads","Tails"])for _ in range(n)]
heads=tosses.count("Heads")
tails=tosses.count("Tails")

p_head= heads /n
p_tails= tails /n
print("number of heads :",heads)
print("number of tails :",tails)
print("Probablity of heads :",p_head)
print("Probablity of tails :",p_tails)
```

```python
#problem 3
import random
from collections import Counter
n = 6000
die_rolls = [random.randint(1, 6) for _ in range(n)]
counts = Counter(die_rolls)

print("\nPROBLEM 3: DIE ROLLED 6000 TIMES")

for outcome in range(1, 7):
    probability = counts[outcome] / n
    print(
        f"Outcome {outcome}: "
        f"Frequency = {counts[outcome]}, "
        f"Experimental Probability = {probability:.4f}"
    )
```

```python
#problem 4
import random
rolls = [random.randint(1, 6) for _ in range(10000)]

A = rolls.count(1)
B = rolls.count(6)

A_union_B = A + B

P_A = A / 10000
P_B = B / 10000
P_A_union_B = A_union_B / 10000

print("Number of rolls:", 10000)
print("Number of 1s (A):", A)
print("Number of 6s (B):", B)

print("\nP(A) =", P_A)
print("P(B) =", P_B)
print("P(A) + P(B) =", P_A + P_B)
print("P(A ∪ B) =", P_A_union_B)

if abs(P_A_union_B - (P_A + P_B)) < 1e-10:
    print("\nVerified: P(A ∪ B) = P(A) + P(B)")
    print("A and B are disjoint events.")
else:
    print("\nThe rule was not verified.")
```

```python
#problem 5
sample_space = {1, 2, 3, 4, 5, 6}
A = {2, 4, 6}
B = {4, 5, 6}

A_union_B = A | B
A_intersection_B = A & B

P_A = len(A) / len(sample_space)
P_B = len(B) / len(sample_space)
P_intersection = len(A_intersection_B) / len(sample_space)
P_union = len(A_union_B) / len(sample_space)

# Display results
print("A =", A)
print("B =", B)
print("A ∪ B =", A_union_B)
print("A ∩ B =", A_intersection_B)

print("\nP(A) =", P_A)
print("P(B) =", P_B)
print("P(A ∩ B) =", P_intersection)
print("P(A ∪ B) =", P_union)

# Verify addition rule
formula = P_A + P_B - P_intersection

print("\nP(A) + P(B) - P(A ∩ B) =", formula)

if P_union == formula:
    print("Verified: P(A ∪ B) = P(A) + P(B) - P(A ∩ B)")
else:
    print("Not verified")
```

```python
#problem 6
import random
n = 10000
heads = 0
even = 0
both = 0

for _ in range(n):
    coin = random.choice(["H", "T"])
    die = random.randint(1, 6)
    if coin == "H":
        heads += 1
    if die % 2 == 0:
        even += 1
    if coin == "H" and die % 2 == 0:
        both += 1

P_A = heads / n
P_B = even / n
P_intersection = both / n

print("Number of trials:", n)
print("Heads:", heads)
print("Even numbers:", even)
print("Head and Even:", both)

print("\nP(A) =", P_A)
print("P(B) =", P_B)
print("P(A ∩ B) =", P_intersection)

product = P_A * P_B

print("P(A) × P(B) =", product)

if abs(P_intersection - product) < 0.01:
    print("\nVerified: P(A ∩ B) ≈ P(A) × P(B)")
    print("A and B are independent events.")
else:
    print("\nThe simulation result differs slightly due to random variation.")
```
